```python
def greet(name):
    print("Hello", name)
    
greet("Prakash")
greet("Omi")
```

```python
def add(a, b):
    print(a+b)

add(10, 20)
```

```python
def student_info(name, age, course):
    print("Name:", name)
    print("Age:", age)
    print("Course:", course)

student_info("Ram", 22, "BCA")
student_info("John", 24, "BSCS")
```

## Keyword Argument

```python
def student(name, age):
    print(name)
    print(age)

# student(22, "Ram")
student(age=22, name="Ram")
```

## Return Statement

```python
def add(a,b):
    return a+b
    
result = add(10, 20)
print(result)
```

```python
def add(a,b):
    return a+b

def sub(a,b):
    return a-b

def mul(a,b):
    return a*b

def div(a,b):
    if b == 0:
        return "Error! Cannot divide by zero."
    return a / b

print("Select Operation: ")
print("1. Addition")
print("2. Subtraction")
print("3. Multiplication")
print("4. Division")

choice = input("Enter your choice (1-4): ")

num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))

if choice == "1":
    print("Result:", add(num1, num2))

elif choice == "2":
    print("Result:", sub(num1, num2))

elif choice == "3":
    print("Result:", mul(num1, num2))

elif choice == "4":
    print("Result:", div(num1, num2))

else:
    print("Invalid choice!")
```
