```python
#enumerate() - Lopp with Index

# Access both index and value

fruits = ["apple", "banana", "cherry"]
for index, fruit in enumerate (fruits):
    print(f"Index{index}:{fruit}")
print("\n")
    
#Start from 1 instead of 0

for index, fruit in enumerate(fruits, start=1):
    print(f"Index{index}:{fruit}")
```

```python
# zip() - Loop Multiple Sequences Together

names = ["Alice", "Bob", "Charlie"]
ages = [24, 30, 35]
cities = ["NYC", "LA", "Chicago"]

for name, age, city in zip(names, ages, cities):
    print(f"{name} is {age} years old and lives in {city}")
```

```python
count = 1

# 2. Set the loop condition
while count <= 5:
    # 3. Code block to execute
    print(f"The count is: {count}")
    
    # 4. Increment the variable to eventually break the loop
    count += 1
```

```python
# Break and Continue in Loops

for i in range(10):
    if i == 5:
        break
    print(i)
    
```

```python
# Search and Stop

numbers = [3, 4, 5, 7, 9, 6, 2]
for num in numbers:
    if num % 2 == 0:
        print(f"Found first even number: {num}")
        break
```

```python
# Continue Statement

for i in range(10):
    if i % 2 == 0:
        continue # Skip even numbers
    print (i)

    # Only odd numbers
```

# Sales Commission Calculator

```python
### print("--- Employee Sales Commission Calculator ---")

total_employee = int(input("Enter how many employees to get commission: "))
for i in range(total_employee):
    
    sales = float(input(f"\nEnter total sales for Employee {total_employee}: $"))
    sales_name = input("Enter the name: ")
    
    if sales >= 50000:
        commission_rate = 0.12  # 12% commission for high sales
    elif sales >= 20000:
        commission_rate = 0.08  
    elif sales >= 10000:
        commission_rate = 0.05  
    # elif sales <= 10000:
    #     commission_rate = 0.02
    else:
        commission_rate = 0.02  # 0% commission for sales below 10,000
        
    # Calculate the commission amount
    commission_earned = sales * commission_rate
    
    # Display the results for the employee
    print(f"Employee {i} Rate: {commission_rate * 100}%")
    print(f"Employee {i} Commission Earned: ${commission_earned:,.2f}")

# print("\n--- All 10 employees processed successfully ---")
```
