```python
num = int(input("Enter a number: "))

if num > 0:
    print("Positive!")
elif num < 0:
    print("Negative!")
else:
    print("ZERO!")
```

```python
marks = float(input("Enter marks:"))

if marks < 0 or marks > 100:
    print("Invalid Number! Try Again.")

if marks >= 80:
    grade = "A"
elif marks >= 70:
    grade = "B"
elif marks >= 60:
    grade = "C"
elif marks >= 50:
    grade = "D"
else:
    grade = "Fail"

print("Grade:", grade)
```
