```python
# Trip Planner

destination = input("Enter destination (Pokhara/Chitwan/Kathmandu): ")
people = int(input("Enter the number of people: "))
nights = int(input("Enter number of nights: "))
accommodation = input("Enter accommodation type (Budget/Standard/Luxury): ")

# Transport rate
if destination == "Pokhara":
    transport_rate = 3000
elif destination == "Chitwan":
    transport_rate = 2500
elif destination == "Kathmandu":
    transport_rate = 1500
else:
    transport_rate = 0
    print("Invalid destination!")

# Accommodation rate
if accommodation == "Budget":
    accommodation_rate = 1000
elif accommodation == "Standard":
    accommodation_rate = 2000
elif accommodation == "Luxury":
    accommodation_rate = 4000
else:
    accommodation_rate = 0
    print("Invalid accommodation type!")

# Calculate costs
transport_cost = transport_rate * people
accommodation_cost = accommodation_rate * people * nights

total_cost = transport_cost + accommodation_cost

# Calculate discount
discount_percent = 0

if people >= 4:
    discount_percent = discount_percent + 10

if nights >= 5:
    discount_percent = discount_percent + 5

if discount_percent > 15:
    discount_percent = 15

discount_amount = total_cost * discount_percent / 100

final_cost = total_cost - discount_amount

cost_per_person = final_cost / people

# Display results
print("\n--- Trip Details ---")
print("Destination:", destination)
print("Number of people:", people)
print("Number of nights:", nights)
print("Transport cost: Rs.", transport_cost)
print("Accommodation cost: Rs.", accommodation_cost)
print("Total cost before discount: Rs.", total_cost)
print("Discount:", discount_percent, "%")
print("Discount amount: Rs.", discount_amount)
print("Final trip cost: Rs.", final_cost)
print("Cost per person: Rs.", cost_per_person)
```

```python
# Cashback Calculator

amount = float(input("Enter transaction amount: Rs. "))
method = input("Enter payment method (App/QR): ")

if amount < 1000:
    print("\nTransaction is not eligible for cashback.")
    print("Cashback amount: Rs. 0")
    print("QR bonus added: No")
    print("Final cashback: Rs. 0")

else:
    cashback = amount * 0.05

    # Maximum cashback is Rs. 200
    if cashback > 200:
        cashback = 200

    # QR bonus
    qr_bonus = 0

    if method == "QR":
        qr_bonus = 20
        bonus_added = "Yes"
    else:
        bonus_added = "No"

    final_cashback = cashback + qr_bonus

    print("\nTransaction is eligible for cashback.")
    print("Cashback amount: Rs.", cashback)
    print("QR bonus added:", bonus_added)
    print("Final cashback: Rs.", final_cashback)
```
